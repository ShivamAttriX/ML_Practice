```python
# Find product names that have Size 9
size_9 = cleaned_df.loc[
    cleaned_df['Size'] == '9', 'Product_Name'
].unique()

# Find product names that have Size M
size_M = cleaned_df.loc[
    cleaned_df['Size'] == 'M', 'Product_Name'
].unique()

# Find common product names
common_products = set(size_9).intersection(set(size_M))

# Display all matching rows
result = cleaned_df[
    (cleaned_df['Product_Name'].isin(common_products)) &
    (cleaned_df['Size'].isin(['9', 'M']))
].sort_values('Product_Name')

display(result)
```

```python
# Restore and preserve original size information

cleaned_df['Size_Raw'] = (
    df['Size']
    .astype('string')
    .str.strip()
    .str.upper()
)

is_numeric = cleaned_df['Size_Raw'].str.fullmatch(
    r'\d+',
    na=False
)

cleaned_df['Size_Type'] = np.select(
    [
        is_numeric,
        cleaned_df['Size_Raw'].isin(['S', 'M', 'L', 'XL', 'XXL'])
    ],
    [
        'Numeric',
        'Alphabetical'
    ],
    default='Unknown'
)

cleaned_df['Size_Numeric'] = pd.to_numeric(
    cleaned_df['Size_Raw'].where(is_numeric),
    errors='coerce'
)

cleaned_df['Size'] = cleaned_df['Size_Raw']
```

```python
# First standardize capitalization, then map the typos
cleaned_df['Region'] = cleaned_df['Region'].str.title().replace({
    'Bengaluru': 'Bangalore',
    'Hyd': 'Hyderabad',
    'Hyderbad': 'Hyderabad'
})
```

```python
# Calculate units sold
calculated_units = (
    cleaned_df['Revenue'] /
    (cleaned_df['MRP'] * (1 - cleaned_df['Discount_Applied']))
)

# Identify valid records
mask = (
    cleaned_df['Units_Sold'].isna() &
    cleaned_df['MRP'].notna() &
    cleaned_df['Discount_Applied'].notna() &
    cleaned_df['Revenue'].notna() &
    (cleaned_df['MRP'] > 0) &
    (cleaned_df['Discount_Applied'] >= 0) &
    (cleaned_df['Discount_Applied'] < 1)
)

# Replace missing Units_Sold
cleaned_df.loc[mask, 'Units_Sold'] = calculated_units[mask].round().astype(int)
```

```python
mask = (
    cleaned_df['MRP'].isna() &
    cleaned_df['Units_Sold'].notna() &
    cleaned_df['Revenue'].notna() &
    cleaned_df['Discount_Applied'].notna() &
    (cleaned_df['Units_Sold'] > 0) &
    (cleaned_df['Discount_Applied'] >= 0) &
    (cleaned_df['Discount_Applied'] < 1)
)

cleaned_df.loc[mask, 'MRP'] = (
    cleaned_df.loc[mask, 'Revenue'] /
    (
        cleaned_df.loc[mask, 'Units_Sold'] *
        (1 - cleaned_df.loc[mask, 'Discount_Applied'])
    )
).round(2)
```

```python
mask = (
    cleaned_df['Discount_Applied'].isna() &
    cleaned_df['Units_Sold'].notna() &
    cleaned_df['MRP'].notna() &
    cleaned_df['Revenue'].notna() &
    (cleaned_df['Units_Sold'] > 0) &
    (cleaned_df['MRP'] > 0)
)

cleaned_df.loc[mask, 'Discount_Applied'] = (
    1 - (
        cleaned_df.loc[mask, 'Revenue'] /
        (
            cleaned_df.loc[mask, 'Units_Sold'] *
            cleaned_df.loc[mask, 'MRP']
        )
    )
).round(4)
```
