```python
# Find product names that have Size 9
size_9 = cleaned_df.loc[
    cleaned_df['Size'] == '9', 'Product_Name'
].unique()

# Find product names that have Size M
size_M = cleaned_df.loc[
    cleaned_df['Size'] == 'M', 'Product_Name'
].unique()

# Find common product names
common_products = set(size_9).intersection(set(size_M))

# Display all matching rows
result = cleaned_df[
    (cleaned_df['Product_Name'].isin(common_products)) &
    (cleaned_df['Size'].isin(['9', 'M']))
].sort_values('Product_Name')

display(result)
```

```python
# Restore and preserve original size information

cleaned_df['Size_Raw'] = (
    df['Size']
    .astype('string')
    .str.strip()
    .str.upper()
)

is_numeric = cleaned_df['Size_Raw'].str.fullmatch(
    r'\d+',
    na=False
)

cleaned_df['Size_Type'] = np.select(
    [
        is_numeric,
        cleaned_df['Size_Raw'].isin(['S', 'M', 'L', 'XL', 'XXL'])
    ],
    [
        'Numeric',
        'Alphabetical'
    ],
    default='Unknown'
)

cleaned_df['Size_Numeric'] = pd.to_numeric(
    cleaned_df['Size_Raw'].where(is_numeric),
    errors='coerce'
)

cleaned_df['Size'] = cleaned_df['Size_Raw']
```

```python
# First standardize capitalization, then map the typos
cleaned_df['Region'] = cleaned_df['Region'].str.title().replace({
    'Bengaluru': 'Bangalore',
    'Hyd': 'Hyderabad',
    'Hyderbad': 'Hyderabad'
})
```

```python
# Calculate units sold
calculated_units = (
    cleaned_df['Revenue'] /
    (cleaned_df['MRP'] * (1 - cleaned_df['Discount_Applied']))
)

# Identify valid records
mask = (
    cleaned_df['Units_Sold'].isna() &
    cleaned_df['MRP'].notna() &
    cleaned_df['Discount_Applied'].notna() &
    cleaned_df['Revenue'].notna() &
    (cleaned_df['MRP'] > 0) &
    (cleaned_df['Discount_Applied'] >= 0) &
    (cleaned_df['Discount_Applied'] < 1)
)

# Replace missing Units_Sold
cleaned_df.loc[mask, 'Units_Sold'] = calculated_units[mask].round().astype(int)
```

```python
mask = (
    cleaned_df['MRP'].isna() &
    cleaned_df['Units_Sold'].notna() &
    cleaned_df['Revenue'].notna() &
    cleaned_df['Discount_Applied'].notna() &
    (cleaned_df['Units_Sold'] > 0) &
    (cleaned_df['Discount_Applied'] >= 0) &
    (cleaned_df['Discount_Applied'] < 1)
)

cleaned_df.loc[mask, 'MRP'] = (
    cleaned_df.loc[mask, 'Revenue'] /
    (
        cleaned_df.loc[mask, 'Units_Sold'] *
        (1 - cleaned_df.loc[mask, 'Discount_Applied'])
    )
).round(2)
```

```python
mask = (
    cleaned_df['Discount_Applied'].isna() &
    cleaned_df['Units_Sold'].notna() &
    cleaned_df['MRP'].notna() &
    cleaned_df['Revenue'].notna() &
    (cleaned_df['Units_Sold'] > 0) &
    (cleaned_df['MRP'] > 0)
)

cleaned_df.loc[mask, 'Discount_Applied'] = (
    1 - (
        cleaned_df.loc[mask, 'Revenue'] /
        (
            cleaned_df.loc[mask, 'Units_Sold'] *
            cleaned_df.loc[mask, 'MRP']
        )
    )
).round(4)
```

```python
# Preserve original dates
cleaned_df['Order_Date_Original'] = df['Order_Date'].copy()

raw_dates = (
    cleaned_df['Order_Date_Original']
    .astype('string')
    .str.strip()
)

parsed_dates = pd.Series(
    pd.NaT,
    index=cleaned_df.index,
    dtype='datetime64[ns]'
)

# YYYY-MM-DD
mask_iso_dash = raw_dates.str.fullmatch(
    r'\d{4}-\d{2}-\d{2}',
    na=False
)

parsed_dates.loc[mask_iso_dash] = pd.to_datetime(
    raw_dates.loc[mask_iso_dash],
    format='%Y-%m-%d',
    errors='coerce'
)

# YYYY/MM/DD
mask_iso_slash = raw_dates.str.fullmatch(
    r'\d{4}/\d{2}/\d{2}',
    na=False
)

parsed_dates.loc[mask_iso_slash] = pd.to_datetime(
    raw_dates.loc[mask_iso_slash],
    format='%Y/%m/%d',
    errors='coerce'
)

# DD-MM-YYYY
mask_dmy = raw_dates.str.fullmatch(
    r'\d{2}-\d{2}-\d{4}',
    na=False
)

# Assumes hyphenated dates use day-first ordering.
# Confirm this convention against the source data.
parsed_dates.loc[mask_dmy] = pd.to_datetime(
    raw_dates.loc[mask_dmy],
    format='%d-%m-%Y',
    errors='coerce'
)

cleaned_df['Order_Date'] = parsed_dates

print("Original missing:", raw_dates.isna().sum())
print("Parsed missing:", parsed_dates.isna().sum())

# Identify non-null strings that failed parsing
unparsed = raw_dates.notna() & parsed_dates.isna()

print("Unparsed records:", unparsed.sum())

display(raw_dates[unparsed].value_counts())
```

```python
categorical_cols = [
    'Gender_Category',
    'Product_Line',
    'Product_Name',
    'Size',
    'Sales_Channel',
    'Region'
]

for col in categorical_cols:
    cleaned_df[col] = cleaned_df[col].astype('string').str.strip()
```
